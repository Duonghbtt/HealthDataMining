```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("check_prescriptions").getOrCreate()

file_path = r"D:\HeKhuyenNghi\HealthDataMining_HKN\data\raw\icu\chartevents.csv.gz"

df = (
    spark.read
    .option("header", True)
    .option("inferSchema", False)
    .csv(file_path)
)
print("Số dòng:", df.count())
print("Số cột:", len(df.columns))
print("Columns:")
print(df.columns)

print("Schema:")
df.printSchema()

print("Sample rows:")
df.show(5, truncate=False)
```
